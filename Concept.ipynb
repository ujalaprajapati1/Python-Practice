# Write a program to find the second-largest number in a list without using sort() or sorted()

```python
numbers = [3,5,8,4,2,6,4,2,5,8,5,6,7,9,1,2,3,4,5,6,7,8,9]
print("Mix Data = ", numbers)
numbers = list(set(numbers))
print("Unique Data = ", numbers)
print("Second largest Number = ", numbers[-2])
```

# Write a program to remove duplicates from a list without using set().

```python
numbers = [3,5,8,4,2,6,4,2,5,8,5,6,7,9,1,2,3,4,5,6,7,8,9]
print("Mix Data = ", numbers)
num = []
for i in numbers:
    if i not in num:
        num.append(i)
        
print("Unique Data = ", num)
```
